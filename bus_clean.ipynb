```python
import numpy as np
import pandas as pd
import math
import string
import re
import geopandas as gpd
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
df = pd.read_csv("/Users/julian/Code/Bus/Bus_Stops_2526.csv")
df.tail(30)
```

```python
def DMS_to_latlong(dms_tag: str):
    dms_split = re.split(r"""[°'"\s]""",dms_tag)
    print(len(dms_split))
    print(dms_split)
    lat_deg = float(dms_split[0].removeprefix('0'))
    lat_min = float(dms_split[1].removeprefix('0'))/60
    lat_sec = float(dms_split[2].removeprefix('0'))/3600
    long_deg = float(dms_split[4].removeprefix('0'))
    long_min = float(dms_split[5].removeprefix('0'))/60
    long_sec = float(dms_split[6].removeprefix('0'))/3600
    lat  = lat_deg + lat_min + lat_sec
    long = long_deg + long_min + long_sec
    return lat, long
```

```python
df.iloc[0,1]
```

```python
AM_standard = []
PM_standard = []
for index, row in df.iterrows():
    stop = row["AM"] 
    if pd.notna(stop):
        if "," not in stop:
            AM_standard.append(DMS_to_latlong(stop))
        else:
            AM_standard.append(tuple([float(x) for x in stop.split(", ")]))
    else:
        AM_standard.append(0)

for index, row in df.iterrows():
    stop = row["PM"] 
    if pd.notna(stop):
        if "," not in stop:
            PM_standard.append(DMS_to_latlong(stop))
        else:
            PM_standard.append(tuple([float(x) for x in stop.split(", ")]))
    else:
        PM_standard.append(0)


```

```python
list = [x[0] for x in AM_standard if not (x==0) ]
list
```

```python
df["AM_lat"] = [x[0] if (not (x==0)) else 0 for x in AM_standard]
df["AM_long"] = [x[1] if (not (x==0)) else 0 for x in AM_standard]


df["PM_lat"] = [x[0] if (not (x==0)) else 0 for x in PM_standard]
df["PM_long"] = [x[1] if (not (x==0)) else 0 for x in PM_standard]

df = df.drop(labels=["AM","PM"],axis=1)
```
